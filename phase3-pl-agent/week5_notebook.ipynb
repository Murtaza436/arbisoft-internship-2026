```python
import os
import sys
import pandas as pd
sys.path.insert(0, '.')
from dotenv import load_dotenv
from src.tools.live_tool import (
    get_standings, get_top_scorers,
    get_recent_results, get_upcoming_fixtures,
    get_team_matches
)
from src.tools.search_tool import brave_search
from src.ingest.pipeline import ingest_statsbomb, get_collection
from src.tools.rag_tool import search_historical_stats
from src.constants import (
    AGENT_COLOR, USER_COLOR, RESET, BOLD,
    INFO_COLOR, SUCCESS_COLOR, ERROR_COLOR, CYAN
)

load_dotenv()
print(f"{CYAN}{BOLD}Week 5 — Premier League Research Agent Scaffold{RESET}")
print(f"{INFO_COLOR}Data Sources:")
print("  1. StatsBomb open data (historical 2015-16 to 2022-23)")
print("  2. football-data.org API (live current season)")
print(f"  3. Brave Search (news and transfers){RESET}\n")
print("All imports successful.")
```

```python
print(f"{USER_COLOR}{BOLD}Testing Data Source 1: StatsBomb Ingestion{RESET}\n")
print(f"{INFO_COLOR}Ingesting 1 season to verify pipeline works...")
print("This takes 5-10 minutes...\n{RESET}")

collection = ingest_statsbomb('data/statsbomb', max_seasons=1)
if collection:
    print(f"\n{SUCCESS_COLOR}{BOLD}Ingestion successful!")
    print(f"Total player-match records: {collection.count()}{RESET}")
else:
    print(f"{ERROR_COLOR}Ingestion failed - check StatsBomb data path{RESET}")
```

```python
print(f"{USER_COLOR}{BOLD}Testing RAG Search on Ingested Data{RESET}\n")

queries = [
    "Jamie Vardy goals Leicester City",
    "Mohamed Salah assists Liverpool",
    "Harry Kane shots Tottenham",
]

for query in queries:
    print(f"{INFO_COLOR}Query: {query}{RESET}")
    result = search_historical_stats(query, n_results=2)
    print(result)
    print()
```
