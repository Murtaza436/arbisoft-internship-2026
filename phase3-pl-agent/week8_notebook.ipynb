# Part 2 — Historical Match Explorer

The OpenFootball dataset has been embedded into **ChromaDB**, allowing semantic retrieval of historical Premier League matches.

Unlike traditional keyword search, the agent understands club names and retrieves relevant fixtures across every available season.

This section demonstrates historical match retrieval.

````python
import pandas as pd
from IPython.display import display

print(f"{CYAN}{BOLD}HISTORICAL MATCH EXPLORER{RESET}\n")

queries = [
    "Liverpool Manchester City",
    "Arsenal Chelsea",
    "Manchester United Liverpool",
]

for query in queries:

    print(f"{USER_COLOR}{BOLD}Fixture:{RESET} {query}")
    print(f"{INFO_COLOR}{'─'*85}{RESET}")

    result = search_historical_stats(
        query,
        n_results=100      # retrieve every meeting
    )

    if isinstance(result, pd.DataFrame):

        display(
            result.style
            .hide(axis="index")
            .set_caption(f"Historical Fixtures — {query}")
            .set_properties(**{
                "text-align":"center",
                "font-size":"11pt"
            })
        )

        print(f"{SUCCESS_COLOR}Total Matches Retrieved: {len(result)}{RESET}")

    else:
        print(result)

    print()
````

# Part 8 — Tool Transparency

Every response generated by the research agent is logged.

These logs reveal:

- Which tools were selected
- In what order they were executed
- The reasoning path taken by the agent

This greatly improves transparency and debugging.

````python
import pandas as pd
from IPython.display import display

logs = get_logs()

tool_calls = []

for log in logs:

    if log["event"] == "tool_call":

        tool_calls.append({

            "Tool": log["tool"],

            "Timestamp": log.get("timestamp", "-"),

            "Status": "Executed"

        })

df = pd.DataFrame(tool_calls)

print(f"{CYAN}{BOLD}TOOLS USED DURING THIS SESSION{RESET}\n")

display(

    df.style

    .hide(axis="index")

    .set_properties(**{

        "text-align":"center"

    })

)
````

# Part 9 — Session Analytics

The notebook now summarizes the entire execution.

Metrics include:

- Number of tool calls
- Tool usage frequency
- Memory usage
- Total logged events

This provides an overview of the complete agent session.

````python
print(f"{CYAN}{BOLD}SESSION SUMMARY{RESET}\n")

logs = get_logs()

tool_counts = {}

for log in logs:
    if log["event"] == "tool_call":
        tool = log["tool"]
        tool_counts[tool] = tool_counts.get(tool, 0) + 1

summary = pd.DataFrame({
    "Tool": tool_counts.keys(),
    "Calls": tool_counts.values()
}).sort_values("Calls", ascending=False)

display(
    summary.style
    .hide(axis="index")
    .set_properties(**{
        "text-align":"center"
    })
)

print()

print(f"{SUCCESS_COLOR}Facts remembered:{RESET}")

# Use memory.facts (the list) instead of memory.get_facts() (the string)
facts_list = memory.facts if memory.facts else ["No facts stored yet."]

display(
    pd.DataFrame({
        "Stored Facts": facts_list
    })
)

print()

print(f"{SUCCESS_COLOR}Total Events Logged:{RESET} {len(logs)}")
````
