## Week 5 Findings

### MCP Server
Built a custom MCP server using the MCP Python SDK that exposes:
- One resource: a local knowledge base text file accessible via URI
- One tool: web_search that searches Brave and returns results

The MCP client connects to the server via stdio transport,
initializes a session, and calls tools or reads resources
using the standard MCP protocol.

### Multi-Agent Architecture
Built a supervisor and worker pattern with two specialized workers:
- search_worker: searches the web and summarizes results
- summary_worker: reads files and summarizes content

The supervisor receives any user request, asks the LLM to decide
which worker is most appropriate, routes the task, and returns
the result. This is called task routing or agent orchestration.

### Tracing Layer
Every event across the entire agent graph is logged to logs/trace.log
with timestamp, agent name, event type, and data. This creates a
complete audit trail showing exactly how each request was handled,
which agents were involved, and in what order.

### Key Concepts
- MCP standardizes AI tool connections like USB standardizes devices
- Supervisor pattern separates routing logic from execution logic
- Workers are specialized and do one thing well
- Tracing enables debugging and monitoring of complex agent systems
